```python
# imports
import pandas as pd
import joblib
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

---

### Load Housing Data

We're only going to use 3 features in this model

* `Gr_Liv_Area`
* `Bedrooms`
* `Bathrooms`

```python
# load in dataset, call it housing
housing = pd.read_csv("../data/ames.csv")

housing.head()
```

```python
housing.columns
```

```python
# only grabbing columns of interest
housing = housing[['Sale_Price', 'Gr_Liv_Area', 'Bedroom_AbvGr', 'Full_Bath']].copy()

# clean up the column names
housing = housing.rename(columns={
    'Sale_Price': 'sale_price',
    'Gr_Liv_Area': 'sqft',
    'Bedroom_AbvGr': 'bedrooms',
    'Full_Bath': 'bathrooms'
})

# call .head() to check our work
housing.head()
```

```python
housing.drop(columns='sale_price')
```

```python
# Create our feature matrix, X, and target vector, y
X = housing.drop(columns='sale_price') 
y = housing['sale_price']

# Train-test-split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
X_train.shape[0] == y_train.shape[0]
```

```python
# Let's make a model and get some predictions
model = LinearRegression()

# fit it
model.fit(X_train, y_train)
```

```python
# Get R2 Score on both training and test set
print(f"Training R2: {model.score(X_train, y_train)}")
print(f"Testing R2: {model.score(X_test, y_test)}")
```

---

## Saving our model

Our scores are close enough on our training and test set. Let's save our model!

[Joblib](https://joblib.readthedocs.io/en/stable/)

Joblib is a set of tools to provide lightweight pipelining in Python. In particular:

transparent disk-caching of functions and lazy re-evaluation (memoize pattern)

easy simple parallel computing

Joblib is optimized to be fast and robust on large data in particular and has specific optimizations for numpy arrays.

```python
# let's save our model
# use the .dump(model, 'filepath') method
joblib.dump(model, "./../models/model.joblib")
```
